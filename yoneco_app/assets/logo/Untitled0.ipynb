1. Read in the data and set it to a dataframe called bike

```r
# Read the CSV file
bike <- read.csv("train.csv")

# Check the structure
str(bike)
```

2. Check the head of df

```r
head(bike)
```

3. The target we are trying to predict is count - the total number of bike rentals (casual + registered).

4. Create a scatter plot of count vs temp

```r
library(ggplot2)

ggplot(bike, aes(x = temp, y = count)) +
  geom_point(alpha = 0.3, color = "blue") +
  labs(title = "Count vs Temperature",
       x = "Temperature (°C)",
       y = "Bike Rental Count") +
  theme_minimal()
```

5. Plot count versus datetime with temperature color gradient

```r
# Convert datetime to POSIXct
bike$datetime <- as.POSIXct(bike$datetime)

ggplot(bike, aes(x = datetime, y = count, color = temp)) +
  geom_point(alpha = 0.5) +
  scale_color_gradient(low = "blue", high = "red") +
  labs(title = "Bike Rentals Over Time",
       x = "Date",
       y = "Bike Rental Count",
       color = "Temperature (°C)") +
  theme_minimal()
```

6. Correlation between temp and count

```r
cor(bike$temp, bike$count)
```

7. Boxplot of count by season

```r
bike$season <- as.factor(bike$season)

ggplot(bike, aes(x = season, y = count, fill = season)) +
  geom_boxplot() +
  labs(title = "Bike Rentals by Season",
       x = "Season (1=Spring, 2=Summer, 3=Fall, 4=Winter)",
       y = "Bike Rental Count") +
  theme_minimal()
```

8. Create an "hour" column

```r
bike$hour <- format(bike$datetime, "%H")
head(bike[c("datetime", "hour")])
```

9. Scatterplot of count vs hour for working days

```r
working_days <- bike[bike$workingday == 1, ]

ggplot(working_days, aes(x = hour, y = count, color = temp)) +
  geom_point(position = position_jitter(w = 1, h = 0), alpha = 0.5) +
  scale_color_gradientn(colors = c("blue", "green", "yellow", "red")) +
  labs(title = "Bike Rentals by Hour (Working Days)",
       x = "Hour of Day",
       y = "Bike Rental Count",
       color = "Temperature (°C)") +
  theme_minimal()
```

10. Scatterplot of count vs hour for non-working days

```r
non_working_days <- bike[bike$workingday == 0, ]

ggplot(non_working_days, aes(x = hour, y = count, color = temp)) +
  geom_point(position = position_jitter(w = 1, h = 0), alpha = 0.5) +
  scale_color_gradientn(colors = c("blue", "green", "yellow", "red")) +
  labs(title = "Bike Rentals by Hour (Non-Working Days)",
       x = "Hour of Day",
       y = "Bike Rental Count",
       color = "Temperature (°C)") +
  theme_minimal()
```

11. Build temp.model

```r
temp.model <- lm(count ~ temp, data = bike)
```

12. Get summary of temp.model
        Interpretation:

Intercept (6.0462): When temperature is 0°C, we expect about 6 bike rentals

Temp coefficient (9.1705): For each 1°C increase in temperature, bike rentals increase by about 9.17



```r
summary(temp.model)
```

13. Predict bike rentals at 25°C

```r
# Method 1: Using coefficients

6.0462 + 9.1705 * 25

# Method 2: Using predict()
new_data <- data.frame(temp = 25)
predict(temp.model, newdata = new_data)
```

14. Convert hour to numeric

```r
bike$hour <- as.numeric(bike$hour)
```

15. Build full model

```r
# Convert hour to factor for proper treatment
bike$hour_factor <- as.factor(bike$hour)

# Build model using all specified features
full.model <- lm(count ~ season + holiday + workingday + weather + temp + humidity + windspeed + hour_factor,
                data = bike)
```

16. Get summary of the model

```r
summary(full.model)
```

17. Model Performance Evaluation



      Did the model perform well on the training data?

The model shows some improvement over the simple temperature-only model:

Multiple R-squared: 0.3793 (vs 0.1556 for temp-only model)

Adjusted R-squared: 0.3762

Residual standard error: 143.2 (vs 166.5 for temp-only model)

What do you think about using a Linear Model on this data?

While the full model performs better than the temperature-only model, a linear regression approach has limitations for this dataset:

Non-linear relationships: The data shows clear seasonal patterns and time-based trends that linear models struggle to capture

Multiple peaks: The hourly patterns show distinct morning and evening peaks that aren't linear

Growth trend: Bike rentals are increasing over time, which creates a non-stationary pattern

R² of 0.38 means the model only explains about 38% of the variance in bike rentals

Better alternatives would be:

Time series models (ARIMA, SARIMA)

Tree-based models (Random Forest, Gradient Boosting)

Neural networks that can capture complex temporal patterns

The linear model provides a good baseline but isn't the optimal choice for this type of time-series data with complex seasonal patterns.
